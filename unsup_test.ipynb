# Variables Melódicas

```python
import pandas as pd

# ------------------------------------------------------------------
# Load melodic-normalized dataset
# ------------------------------------------------------------------

DATA_PATH = r"C:\Users\Aura De La Garza G\Projects\GenAI-Music\unsupervised\curated_data\measure_features_melodic_normalized.csv"

df = pd.read_csv(DATA_PATH)

print("Dataset shape:", df.shape)


# ------------------------------------------------------------------
# 1. Helper indicators overview
# ------------------------------------------------------------------

print("\n=== Melodic activity indicators ===\n")

display(
    df[["has_melodic_activity", "has_melodic_motion"]]
    .value_counts()
    .rename("count")
    .reset_index()
)


# ------------------------------------------------------------------
# 2. Zeros explained by melodic activity
# ------------------------------------------------------------------

print("\n=== melodic_range == 0 by melodic activity ===\n")

display(
    pd.crosstab(
        df["melodic_range"] == 0,
        df["has_melodic_motion"],
        rownames=["melodic_range == 0"],
        colnames=["has_melodic_motion"],
        margins=True
    )
)


print("\n=== mean_interval == 0 by melodic activity ===\n")

display(
    pd.crosstab(
        df["mean_interval"] == 0,
        df["has_melodic_motion"],
        rownames=["mean_interval == 0"],
        colnames=["has_melodic_motion"],
        margins=True
    )
)


# ------------------------------------------------------------------
# 3. Duration variables conditioned on activity
# ------------------------------------------------------------------

print("\n=== avg_duration distribution (active measures only) ===\n")

display(
    df.loc[df["has_melodic_activity"], "avg_duration"]
    .describe()
    .to_frame()
)


print("\n=== most_common_duration (active measures only) ===\n")

display(
    df.loc[df["has_melodic_activity"], "most_common_duration"]
    .value_counts()
    .head(10)
    .rename("count")
    .reset_index()
)


# ------------------------------------------------------------------
# 4. Large leap percentage sanity check
# ------------------------------------------------------------------

print("\n=== large_jump_pct == 0 interpretation ===\n")

display(
    pd.crosstab(
        df["large_jump_pct"] == 0,
        df["has_melodic_motion"],
        rownames=["large_jump_pct == 0"],
        colnames=["has_melodic_motion"],
        margins=True
    )
)


# ------------------------------------------------------------------
# 5. Single-note vs multi-note measures
# ------------------------------------------------------------------

print("\n=== num_notes_melody distribution ===\n")

display(
    df["num_notes_melody"]
    .value_counts()
    .sort_index()
    .rename("count")
    .reset_index()
)
```

```python
import pandas as pd

# ------------------------------------------------------------------
# Load conditioned dataset
# ------------------------------------------------------------------

DATA_PATH = (
    r"C:\Users\Aura De La Garza G\Projects\GenAI-Music\unsupervised\curated_data"
    r"\measure_features_conditioned.csv"
)

df = pd.read_csv(DATA_PATH)

print("Dataset shape:", df.shape)

# ------------------------------------------------------------------
# 1. Missing values check
# ------------------------------------------------------------------

print("\n=== Remaining missing values (should be zero) ===\n")
display(
    df.isna()
      .sum()
      .sort_values(ascending=False)
      .to_frame("missing_count")
)

# ------------------------------------------------------------------
# 2. Harmonic conditioning validation
# ------------------------------------------------------------------

print("\n=== Harmonic features when no explicit harmony ===\n")

harmonic_numeric = ["degree_relative_to_key", "num_distinct_chords"]

display(
    df.loc[~df["has_explicit_harmony"], harmonic_numeric]
      .describe()
)

print("\n=== Harmonic categorical placeholders ===\n")

display(
    df[["most_common_chord_root", "most_common_chord_type", "function_coarse"]]
    .value_counts()
    .rename("count")
    .reset_index()
    .head(10)
)

# ------------------------------------------------------------------
# 3. Melodic conditioning validation
# ------------------------------------------------------------------

print("\n=== Melodic motion features when no motion ===\n")

melodic_motion_features = ["melodic_range", "mean_interval", "large_jump_pct"]

display(
    df.loc[~df["has_melodic_motion"], melodic_motion_features]
      .describe()
)

print("\n=== Duration features when no melodic activity ===\n")

duration_features = ["avg_duration", "most_common_duration"]

display(
    df.loc[~df["has_melodic_activity"], duration_features]
      .describe()
)

# ------------------------------------------------------------------
# 4. Rhythmic entropy stability
# ------------------------------------------------------------------

print("\n=== Rhythmic entropy check ===\n")
print("Minimum rhythmic entropy:", df["rhythmic_entropy"].min())

display(
    df["rhythmic_entropy"]
    .describe()
    .to_frame()
)

# ------------------------------------------------------------------
# 5. Binary indicators distribution
# ------------------------------------------------------------------

print("\n=== Helper indicator distributions ===\n")

display(
    df[
        [
            "has_explicit_harmony",
            "has_melodic_activity",
            "has_melodic_motion",
        ]
    ]
    .value_counts()
    .rename("count")
    .reset_index()
)

# ------------------------------------------------------------------
# 6. Pitch-class sparsity sanity check
# ------------------------------------------------------------------

print("\n=== Pitch-class sparsity (mean values) ===\n")

pc_cols = [c for c in df.columns if c.startswith("pc_")]

display(
    df[pc_cols]
    .mean()
    .sort_values(ascending=False)
    .to_frame("mean_value")
)

print("\n✔ All checks complete. Dataset is conditioned and stable.")
```

# EDA

```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

DATA_PATH = r"C:\Users\Aura De La Garza G\Projects\GenAI-Music\unsupervised\curated_data\measure_features_conditioned.csv"
df = pd.read_csv(DATA_PATH)

px.defaults.template = "plotly_white"
```

## 2. Harmonic function distribution

```python
counts = df["function_coarse"].value_counts().rename_axis("function_coarse").reset_index(name="count")
fig = px.bar(
    counts,
    x="function_coarse",
    y="count",
    labels={"function_coarse": "Harmonic Function", "count": "Count"},
    title="Distribution of Coarse Harmonic Functions",
)

fig.show()
```

```python
fig = px.histogram(
    df,
    x="function_coarse",
    color="movement",
    barmode="group",
    title="Harmonic Function Distribution by Movement Type"
)

fig.show()
```

```python
melodic_counts = (
    df[["has_melodic_activity", "has_melodic_motion"]]
    .value_counts()
    .reset_index(name="count")
)

fig = px.bar(
    melodic_counts,
    x="count",
    y=["has_melodic_activity", "has_melodic_motion"],
    orientation="h",
    title="Melodic Activity vs Melodic Motion"
)

fig.show()
```

```python
df_motion = df[df["has_melodic_motion"]]

fig = px.scatter(
    df_motion,
    x="melodic_range",
    y="mean_interval",
    color="movement",
    opacity=0.5,
    title="Melodic Gesture Space (Range vs Mean Interval)",
)

fig.show()
```

```python
fig = px.histogram(
    df,
    x="rhythmic_entropy",
    nbins=40,
    title="Distribution of Rhythmic Entropy (Per Measure)"
)

fig.show()
```
