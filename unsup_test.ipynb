# <font color='cyan'>Condicionamento para Completitud de Variables</font>

```python
import pandas as pd

# ------------------------------------------------------------------
# Load conditioned dataset
# ------------------------------------------------------------------

DATA_PATH = (
    r"C:\Users\Aura De La Garza G\Projects\GenAI-Music\unsupervised\curated_data"
    r"\measure_features_conditioned.csv"
)

df = pd.read_csv(DATA_PATH)

print("Dataset shape:", df.shape)

# ------------------------------------------------------------------
# 1. Missing values check
# ------------------------------------------------------------------

print("\n=== Remaining missing values (should be zero) ===\n")
display(
    df.isna()
      .sum()
      .sort_values(ascending=False)
      .to_frame("missing_count")
)

# ------------------------------------------------------------------
# 2. Harmonic conditioning validation
# ------------------------------------------------------------------

print("\n=== Harmonic features when no explicit harmony ===\n")

harmonic_numeric = ["degree_relative_to_key", "num_distinct_chords"]

display(
    df.loc[~df["has_explicit_harmony"], harmonic_numeric]
      .describe()
)

print("\n=== Harmonic categorical placeholders ===\n")

display(
    df[["most_common_chord_root", "most_common_chord_type", "function_coarse"]]
    .value_counts()
    .rename("count")
    .reset_index()
    .head(10)
)

# ------------------------------------------------------------------
# 3. Melodic conditioning validation
# ------------------------------------------------------------------

print("\n=== Melodic motion features when no motion ===\n")

melodic_motion_features = ["melodic_range", "mean_interval", "large_jump_pct"]

display(
    df.loc[~df["has_melodic_motion"], melodic_motion_features]
      .describe()
)

print("\n=== Duration features when no melodic activity ===\n")

duration_features = ["avg_duration", "most_common_duration"]

display(
    df.loc[~df["has_melodic_activity"], duration_features]
      .describe()
)

# ------------------------------------------------------------------
# 4. Rhythmic entropy stability
# ------------------------------------------------------------------

print("\n=== Rhythmic entropy check ===\n")
print("Minimum rhythmic entropy:", df["rhythmic_entropy"].min())

display(
    df["rhythmic_entropy"]
    .describe()
    .to_frame()
)

# ------------------------------------------------------------------
# 5. Binary indicators distribution
# ------------------------------------------------------------------

print("\n=== Helper indicator distributions ===\n")

display(
    df[
        [
            "has_explicit_harmony",
            "has_melodic_activity",
            "has_melodic_motion",
        ]
    ]
    .value_counts()
    .rename("count")
    .reset_index()
)

# ------------------------------------------------------------------
# 6. Pitch-class sparsity sanity check
# ------------------------------------------------------------------

print("\n=== Pitch-class sparsity (mean values) ===\n")

pc_cols = [c for c in df.columns if c.startswith("pc_")]

display(
    df[pc_cols]
    .mean()
    .sort_values(ascending=False)
    .to_frame("mean_value")
)

print("\n✔ All checks complete. Dataset is conditioned and stable.")
```

```python
def eda_musicological_summary(df):
    summary = {}

    # 1. Measure typology
    activity_cols = [
        "has_explicit_harmony",
        "has_melodic_activity",
        "has_melodic_motion"
    ]

    summary["measure_typology"] = (
        df[activity_cols]
        .value_counts(normalize=True)
        .rename("proportion")
        .reset_index()
    )

    # 2. Harmonic function distribution
    summary["harmonic_function_distribution"] = (
        df["function_coarse"]
        .value_counts(normalize=True)
        .rename("proportion")
        .reset_index()
    )

    # 3. Harmonic function by movement (FIXED)
    function_by_movement = (
        df
        .groupby(["movement", "function_coarse"])
        .size()
        .rename("count")
        .reset_index()
    )

    function_by_movement["proportion"] = (
        function_by_movement["count"]
        / function_by_movement.groupby("movement")["count"].transform("sum")
    )

    summary["harmonic_function_by_movement"] = function_by_movement

    # 4. Melodic gesture statistics
    summary["melodic_gesture_stats"] = (
        df[df["has_melodic_motion"]][
            ["melodic_range", "mean_interval", "large_jump_pct"]
        ]
        .describe(percentiles=[0.25, 0.5, 0.75])
    )

    # 5. Rhythmic entropy
    summary["rhythmic_entropy_stats"] = (
        df["rhythmic_entropy"]
        .describe(percentiles=[0.25, 0.5, 0.75])
    )

    summary["rhythmic_entropy_by_movement"] = (
        df
        .groupby("movement")["rhythmic_entropy"]
        .agg(["mean", "median", "std"])
        .sort_values("mean", ascending=False)
    )

    # 6. Pitch-class profile
    pc_cols = [f"pc_{i}" for i in range(12)]

    summary["global_pitch_class_profile"] = (
        df[pc_cols].mean().sort_values(ascending=False)
    )

    pc_sparsity = (df[pc_cols] > 0).sum(axis=1)
    summary["pc_sparsity_stats"] = pc_sparsity.describe()

    return summary
```

```python
eda_musicological_summary = eda_musicological_summary(df)
print(eda_musicological_summary)
```

# <font color='cyan'>Número Óptimo de Clusters</font>

## Prueba de la Silueta y del Codo

```python
import plotly.graph_objects as go
from sklearn.preprocessing import StandardScaler, RobustScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import numpy as np
import pandas as pd
from pathlib import Path

# ------------------------------------------------------------------
# Load dataset
# ------------------------------------------------------------------
IN_PATH = Path(
    r"C:\Users\Aura De La Garza G\Projects\GenAI-Music\unsupervised\curated_data"
    r"\measure_features_conditioned.csv"
)
df = pd.read_csv(IN_PATH)

# ------------------------------------------------------------------
# Numeric features for clustering
# ------------------------------------------------------------------
numeric_features = [
    'melodic_range','mean_interval','large_jump_pct','avg_duration',
    'note_density','rhythmic_entropy','degree_relative_to_key','num_distinct_chords',
] + [f'pc_{i}' for i in range(12)]
X = df[numeric_features]

# ------------------------------------------------------------------
# Scaling
# ------------------------------------------------------------------
z_scores = np.abs((X - X.mean()) / X.std())
outlier_fraction = (z_scores > 3).any(axis=1).mean()
scaler = RobustScaler() if outlier_fraction > 0.05 else StandardScaler()
X_scaled = scaler.fit_transform(X)

# ------------------------------------------------------------------
# PCA (retain 90% variance)
# ------------------------------------------------------------------
pca = PCA(n_components=0.9, random_state=42)
X_pca = pca.fit_transform(X_scaled)

# ------------------------------------------------------------------
# Determine optimal number of clusters
# ------------------------------------------------------------------
k_range = range(2, 15)  # adjust the range if needed
inertia = []
silhouette = []

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(X_pca)
    inertia.append(kmeans.inertia_)
    silhouette.append(silhouette_score(X_pca, labels))

# ------------------------------------------------------------------
# Interactive Elbow Plot
# ------------------------------------------------------------------
fig_elbow = go.Figure()
fig_elbow.add_trace(go.Scatter(
    x=list(k_range),
    y=inertia,
    mode='lines+markers',
    name='Inertia',
    marker=dict(color='blue', size=10)
))
fig_elbow.update_layout(
    title='Elbow Method for Optimal k',
    xaxis_title='Number of Clusters (k)',
    yaxis_title='Inertia',
    template='plotly_white'
)
fig_elbow.show()

# ------------------------------------------------------------------
# Interactive Silhouette Plot
# ------------------------------------------------------------------
fig_sil = go.Figure()
fig_sil.add_trace(go.Scatter(
    x=list(k_range),
    y=silhouette,
    mode='lines+markers',
    name='Silhouette Score',
    marker=dict(color='green', size=10)
))
fig_sil.update_layout(
    title='Silhouette Score for Different k',
    xaxis_title='Number of Clusters (k)',
    yaxis_title='Silhouette Score',
    template='plotly_white'
)
fig_sil.show()
```
